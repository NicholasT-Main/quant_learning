# Ideas
to create a yield prediction model it would be important to have as much data on each specific crop as we can, this is not possible at scale, thus we need to prioritise data based on acesability and its impact on crop yield.
this depends on the crop

## impact to yield
- water availability
- rainfall (depending on farming method)
- nutrient availability
- sunlight
- pest and disease impact
- harvest method
- harvest date

## crop types
as satalite imaging will be used it is important that the main shortfall of satalite imaging is mitigated, this is resolution, thus we require crops farmed on a large scale and as a monoculture.
- field crops
    - corn
    - wheat
    - potatoes
    - barley
- tree yields
    - pecan nuts
    - macadamia nuts
    - avocados
    - apples
- specialty crops
    - lettuce
    - pumpkins
    - melons

The 2 types of crops that would be best to work on with will be field crops or trees due to the lower requirement of resolution
field crop:
pros
- minimised resolution issues
- data available through commodity markets
- more research available
