- nutrient availability
- sunlight

## crop types
as satellite imaging will be used it is important that the main shortfall of satellite imaging is mitigated, this is resolution, thus we require crops farmed on a large scale and as a monoculture.
- field crops
    - corn
    - wheat
    - potatoes
    - barley
    - sugarcane
- tree yields
    - pecan nuts
    - macadamia nuts
    - avocados
    - apples
- specialty crops
    - lettuce
    - pumpkins
    - melons

The 2 types of crops that would be best to work on with will be field crops or trees due to the lower requirement of resolution

field crop:
pros:
- minimised resolution issues
- data available through commodity markets
- more research available
- more likely to be rain-watered (creating more reliable data)
- seasonal crops and yields

cons:
- land is often used for more than one crop and will require satellite crop classification (not always the case)
- grains can store for a long time meaning a crop doesn't directly correlate with price
- some crops can be watered meaning that needs to be predetermined

tree yields:
pros:
- long horizon, meaning potential yield won't fluctuate drastically
- more profitable crops
- possible (concept proven on coffee)
- fruit sales are near harvest (directly affects price)
- harvest dates are longer (yield prediction is more important)
- larger single trees means stress signals will be more visable at lower resoloutions

cons:
- often watered meaning less correlation with rainfall
